```python
import os
import shutil
import pickle
```

```python
train_dir = "../mmaction2/data/table_tennis/train"
val_dir = "../mmaction2/data/table_tennis/val"
test_dir = "../mmaction2/data/table_tennis/test"

mmaction2_ann_file_train = "../mmaction2/data/table_tennis/table_tennis_train.txt"
mmaction2_ann_file_val = "../mmaction2/data/table_tennis/table_tennis_val.txt"
mmaction2_ann_file_test = "../mmaction2/data/table_tennis/table_tennis_test.txt"
```

```python
ntu_train_dir = "../mmaction2/data/table_tennis/ntu_train"
ntu_val_dir = "../mmaction2/data/table_tennis/ntu_val"
ntu_test_dir = "../mmaction2/data/table_tennis/ntu_test"

os.makedirs(ntu_train_dir, exist_ok=True)
os.makedirs(ntu_val_dir, exist_ok=True)
os.makedirs(ntu_test_dir, exist_ok=True)
```

```python
def change_clips_to_ntu_pickle(ann_filepath, original_data_dir, skeleton_data_dir):
    with open(ann_filepath, "r") as f:
        lines = f.readlines()
    count = 0
    for line in lines:

        filename, label = line.strip().split()
        label = int(label) + 1 # as during pose extration, label - 1 is done

        clip_path = original_data_dir + "/" + filename
        ntu_format_filename = f"{filename.split('.')[0]}A{label:03d}.mp4"
        new_path = skeleton_data_dir + "/" + ntu_format_filename

        shutil.copy(clip_path, new_path)
        count += 1
    
    print(f"Formatted {count} clips to NTU naming convention for {ann_filepath}")

change_clips_to_ntu_pickle(mmaction2_ann_file_train, train_dir, ntu_train_dir)
change_clips_to_ntu_pickle(mmaction2_ann_file_val, val_dir, ntu_val_dir)
change_clips_to_ntu_pickle(mmaction2_ann_file_test, test_dir, ntu_test_dir)
```
