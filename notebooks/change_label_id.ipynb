```python
import os

target_file_train = "../mmaction2/data/table_tennis/test_general/table_tennis_train.txt"
target_file_val = "../mmaction2/data/table_tennis/test_general/table_tennis_val.txt"

def change_label(filename):

    lines = []
    with open(filename, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            fname = line.split()[0]
            lines.append(f"{fname} 421\n")

    with open(filename, "w") as f:
        f.writelines(lines)

change_label(target_file_train)
change_label(target_file_val)
```
