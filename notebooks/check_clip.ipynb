```python
import os

train_dir = "../mmaction2/data/table_tennis/train"
val_dir = "../mmaction2/data/table_tennis/val"
test_dir = "../mmaction2/data/table_tennis/test"

mmaction2_ann_file_train = "../mmaction2/data/table_tennis/table_tennis_train.txt"
mmaction2_ann_file_val = "../mmaction2/data/table_tennis/table_tennis_val.txt"
mmaction2_ann_file_test = "../mmaction2/data/table_tennis/table_tennis_test.txt"

def check_missing_videos(annotation_file, video_dir):
    print(f"\nChecking: {annotation_file}")
    missing = []

    with open(annotation_file, "r") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue

            filename, label = line.split()

            full_path = os.path.join(video_dir, filename)

            if not os.path.exists(full_path):
                print("MISSING FILE:", full_path)
                missing.append(full_path)

    print("Total missing files:", len(missing))
    return missing


# Run checks
missing_train = check_missing_videos(mmaction2_ann_file_train, train_dir)
missing_val = check_missing_videos(mmaction2_ann_file_val, val_dir)
missing_test = check_missing_videos(mmaction2_ann_file_test, test_dir)
```
