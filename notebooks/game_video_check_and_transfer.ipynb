```python
import pandas as pd
import os
import paramiko
```

```python
# Directory containing your Excel files
directory = "../data/videos/"
non_missing_video_ids = [413, 437, 439, 440, 444, 445, 452, 453, 457, 461, 462, 465, 466, 467, 468, 469, 472, 473, 475, 476, 477, 483, 485, 486, 487, 489, 490, 491, 492, 493, 494, 497, 501, 502, 504, 505, 506, 508, 513, 520, 521, 527, 528, 529, 530, 531, 533, 535, 536, 537, 538, 541, 542, 543, 544, 545, 546, 547, 548, 549, 550, 551, 552, 553, 555, 556, 560, 561, 562, 563, 564, 565, 566, 569, 570, 571, 575, 576, 577, 598, 599, 600, 601, 603, 605, 609, 610, 620, 621, 622, 623, 625, 12914, 12918, 12920, 12924, 12933, 12941, 12942, 12967, 12971, 12972, 12973, 12978, 12979, 12980, 12981, 12986, 12988, 12992, 12993, 12994, 12998]

results = []

game_xlsx = "../table_tennis_data/metadata/game.xlsx"
df_game = pd.read_excel(game_xlsx, sheet_name="game")
    
# Filter rows where Id is in non_missing_video_ids
filtered_games = df_game[df_game["Id"].isin(non_missing_video_ids)][["Id", "Name"]]
game_list = list(filtered_games.itertuples(index=False, name=None))

for game_id, name in game_list:
    name = name + ".mp4"
    file_path = os.path.join(directory, name)

    if os.path.isfile(file_path):
        results.append((game_id, name))

print(results)
```

```python
# SSH server details
ssh_host = "ai.fisk.edu"
ssh_port = 22
ssh_username = ""
ssh_password = ""
remote_directory = ""

# Create SSH + SFTP connection
ssh = paramiko.SSHClient()
ssh.set_missing_host_key_policy(paramiko.AutoAddPolicy())
ssh.connect(
    hostname=ssh_host,
    port=ssh_port,
    username=ssh_username,
    password=ssh_password
)

sftp = ssh.open_sftp()
for game_id, name in results:
    local_path = os.path.join(directory, name)
    remote_path = os.path.join(remote_directory, name)

    if os.path.isfile(local_path):
        print(f"Uploading {name}...")
        sftp.put(local_path, remote_path)
    else:
        print(f"File not found: {local_path}")

sftp.close()
ssh.close()
```
