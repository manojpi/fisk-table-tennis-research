```python
import os
import pickle
```

```python
skeleton_train_dir = "../mmaction2/data/table_tennis/skeleton_train"
skeleton_val_dir = "../mmaction2/data/table_tennis/skeleton_val"
skeleton_test_dir = "../mmaction2/data/table_tennis/skeleton_test"
```

```python
# convert all the pickle to single pickle

input_dir = "path/to/pkl_clips"   # directory containing many clip_xxxx.pkl

skeleton_train_pickle = "../mmaction2/data/table_tennis/table_tennis_train.pkl"
skeleton_val_pickle = "../mmaction2/data/table_tennis/table_tennis_val.pkl"
skeleton_test_pickle = "../mmaction2/data/table_tennis/table_tennis_test.pkl"


def combine_pickles(input_dir, output_file):
    
    data_list = []

    for fname in os.listdir(input_dir):
        if fname.endswith(".pkl"):
            fpath = os.path.join(input_dir, fname)

            with open(fpath, "rb") as f:
                sample = pickle.load(f)

            data_list.append(sample)

    print("Total samples:", len(data_list))

    # Save merged file
    with open(output_file, "wb") as f:
        pickle.dump(data_list, f)

    print(f"Merged annotation saved to {output_file}")

combine_pickles(skeleton_train_dir, skeleton_train_pickle)
combine_pickles(skeleton_test_dir, skeleton_test_pickle)
combine_pickles(skeleton_val_dir, skeleton_val_pickle)
```
